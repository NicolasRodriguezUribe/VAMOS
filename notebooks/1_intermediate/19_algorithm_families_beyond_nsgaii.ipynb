# Algorithm Families Beyond NSGA-II

VAMOS ships several MOEA families beyond NSGA-II. This notebook gives you short, current recipes for the algorithms that users often skip in first-pass tutorials.


```python
import pandas as pd

from vamos import optimize

CASES = [
    ("spea2", "zdt1", {"max_evaluations": 8000, "pop_size": 100}),
    ("ibea", "zdt1", {"max_evaluations": 8000, "pop_size": 100}),
    ("smpso", "zdt1", {"max_evaluations": 8000, "pop_size": 100}),
    ("nsgaiii", "dtlz2", {"max_evaluations": 12000, "pop_size": 92, "n_obj": 3}),
    ("agemoea", "dtlz2", {"max_evaluations": 12000, "pop_size": 91, "n_obj": 3}),
    ("rvea", "dtlz2", {"max_evaluations": 12000, "pop_size": 91, "n_obj": 3}),
]

rows = []
for algorithm, problem, kwargs in CASES:
    result = optimize(problem, algorithm=algorithm, seed=42, engine="numpy", **kwargs)
    rows.append(
        {
            "algorithm": algorithm,
            "problem": problem,
            "objectives": result.F.shape[1],
            "solutions": len(result),
        }
    )

pd.DataFrame(rows)
```

## What each family is good at

- **SPEA2**: strength-based environmental selection.
- **IBEA**: indicator-based ranking with explicit quality preference.
- **SMPSO**: particle-swarm style search for continuous bi-objective problems.
- **NSGA-III**: reference-direction survival for many-objective problems.
- **AGE-MOEA**: geometry-adaptive many-objective search.
- **RVEA**: reference-vector guidance with angle-penalized distance.

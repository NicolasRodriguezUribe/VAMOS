# Multi-Criteria Decision Making (MCDM)

Pareto optimization returns a *set* of trade-off solutions. MCDM methods help you pick one solution based on the decision-maker's preference model.


```python
import matplotlib.pyplot as plt
import numpy as np

from vamos import optimize
from vamos.ux.api import (
    knee_point_scores,
    reference_point_scores,
    tchebycheff_scores,
    weighted_sum_scores,
)

result = optimize("zdt1", algorithm="nsgaii", max_evaluations=12000, pop_size=100, seed=42)
F = result.F
print(f"Returned {len(F)} non-dominated solutions")
```

```python
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(F[:, 0], F[:, 1], s=30, alpha=0.7, c="steelblue")
ax.set_xlabel("f1")
ax.set_ylabel("f2")
ax.set_title("ZDT1 Pareto front")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## Weighted-sum and Tchebycheff scores


```python
weights = np.array([0.7, 0.3])
ws = weighted_sum_scores(F, weights)
tch = tchebycheff_scores(F, weights)

print("Weighted-sum best:", ws.best_index, ws.best_point)
print("Tchebycheff best:", tch.best_index, tch.best_point)
```

## Preference-point and knee-point scores


```python
ref = np.array([0.25, 0.55])
ref_scores = reference_point_scores(F, ref)
knee = knee_point_scores(F)

print("Reference-point best:", ref_scores.best_index, ref_scores.best_point)
print("Knee-point best:", knee.best_index, knee.best_point)
```

```python
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(F[:, 0], F[:, 1], s=24, alpha=0.55, label="Pareto set")
ax.scatter(*ws.best_point, s=90, marker="*", label="Weighted sum")
ax.scatter(*tch.best_point, s=90, marker="P", label="Tchebycheff")
ax.scatter(*ref_scores.best_point, s=90, marker="X", label="Reference point")
ax.scatter(*knee.best_point, s=90, marker="D", label="Knee")
ax.scatter(*ref, s=70, marker="o", facecolors="none", edgecolors="black", label="Preference point")
ax.set_xlabel("f1")
ax.set_ylabel("f2")
ax.set_title("MCDM choices on the same Pareto set")
ax.legend(loc="best")
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

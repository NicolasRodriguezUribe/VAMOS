# Ablation Planning

This notebook plans scientifically distinct variants through the canonical StudySpec resolver. Each variant becomes its own durable study so its configuration is part of task identity.

```python
from vamos import StudySpec, plan_study

variants = {"baseline": 50, "tuned": 80}
plans = {}
for name, population_size in variants.items():
    spec = StudySpec(
        problems=["zdt1", "dtlz2"],
        algorithms=["nsgaii"],
        seeds=[1, 2, 3],
        max_evaluations=20000,
        pop_size=population_size,
        algorithm_configs={"nsgaii": {"pop_size": population_size}},
        labels={"workflow": "ablation", "variant": name},
    )
    plans[name] = plan_study(spec)

{name: report.plan.task_count for name, report in plans.items()}
```

```python
{name: report.plan.tasks[:3] for name, report in plans.items()}
```

## Interpreting contributions
Given final metrics per variant (e.g., median HV at the full budget), compute deltas vs the baseline.
The largest negative delta when a component is removed indicates the biggest contribution.

```python
metrics = {
    "baseline": 0.55,
    "tuned": 0.67,
}

baseline = metrics["baseline"]
deltas = {name: score - baseline for name, score in metrics.items()}
deltas
```
